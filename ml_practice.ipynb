```python
import pandas as pd 
import numpy as np 
```

Predict the prices of the properties based on the features available to us

```python
#data available with us
data = {
    "area_sqft": [500, 700, 800, 1000, 1200, 1500, 1800, 2000],
    "bedrooms":  [1,   1,   2,    2,    3,    3,    4,    4],
    "price":    [50,  65,  75,   90,   120,  150,  180,  210]  # in lakhs
}

df = pd.DataFrame(data)

df.head()
```

Now our input x variables are area_sqft and bedrooms and our target variable y is price 

```python
#shuffle data 
df = df.sample(frac=1,random_state=42).reset_index(drop=True)

#train test split 
split_index = int(0.75 * len(df))

#75 % train 25 % test data split
train_df = df.iloc[:split_index]
test_df = df.iloc[split_index:]

X_train = train_df[["area_sqft","bedrooms"]]
y_train = train_df["price"]

X_test = test_df[["area_sqft","bedrooms"]]
y_test = test_df["price"]

```

```python
#Fitting a linear regression model 
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train,y_train)
```

```python
#Intercept part
print("Intercept:",model.intercept_)
print("Coefficient:",model.coef_)
```

```python
y_pred = model.predict(X_test)

result = pd.DataFrame({
"Actual":y_test.values,
"Predicted":y_pred
})

result
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test,y_pred)
print("MAE :",mae)
```

```python
data = {
    "area_sqft": [500, 700, 800, 1000, 1200, 1500, 1800, 2000],
    "bedrooms":  [1,   1,   2,    2,    3,    3,    4,    4],
    "price":    [50,  65,  75,   90,   120,  150,  180,  210]  # in lakhs
}

df = pd.DataFrame(data)

df.head()
```

```python
results = pd.DataFrame({
    "Actual":y_test,
    "Predicted":y_pred
})

results
```

```python
#Data splitting logic
df = df.sample(frac=1,random_state=42).reset_index(drop=True)

split_index = int(0.8 * len(df))

df_train = df.iloc[:split_index]

df_test = df.iloc[split_index:]

X_train = df_train[["area_sqft","bedrooms"]]
y_train = df_train["price"]

X_test = df_test[["area_sqft","bedrooms"]]
y_test = df_test["price"]

```

```python
#Initializing the model

model = LinearRegression()

model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)

y_pred 
```

```python
model.intercept_
model.coef_
```

```python
results = pd.DataFrame({
    "Actual":y_test.values,
    "Predicted":y_pred
})

results 
```

```python
mae = mean_absolute_error(y_test,y_pred)

print(f"Mean abosulute error of the model is : {mae}")
```
